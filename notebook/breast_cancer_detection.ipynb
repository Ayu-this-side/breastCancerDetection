# Clean the data

We have 2 columns which is not important for model traning so we will remove them

```python
df.drop(columns = ['Unnamed: 32', 'id'], inplace = True)
```

```python
df.info()
```

```python
# turn target variable into 1s and 0s
df['diagnosis'] = df['diagnosis'].map({'M': 1, 'B': 0})
```

```python
df['diagnosis'].value_counts().plot(kind = 'bar', title = 'Diagnosis Count', xlabel = 'Diagnosis', ylabel = 'Count')
```

# Dividing the input and output columns

```python
X = df.drop(columns = ['diagnosis'])
y = df['diagnosis']
```

## Normalize the data

```python
from sklearn.preprocessing import StandardScaler

# Create a scaler object
scaler = StandardScaler()

# Fit the scaler to the data and transform the data
X_scaled = scaler.fit_transform(X)
```

## Train test split

```python
from sklearn.model_selection import train_test_split
X_train , X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.3, random_state=42)
```

## Traning the model

```python
from sklearn.linear_model import LogisticRegression

# Create logistic regression model
lr = LogisticRegression()

# Train the model on the training data
lr.fit(X_train, y_train)

# Predict the target variable on the test data
y_pred = lr.predict(X_test)
```

## Evaluate the model

```python
from sklearn.metrics import accuracy_score

# Evaluate the accuracy of the model
accuracy = accuracy_score(y_test, y_pred)
print(f'Accuracy: {accuracy:.2f}')
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```
